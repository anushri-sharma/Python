###Resursion : The process in which function calls itself directly or indirectly is called "recursion" and the corresponding function in called "recursive function".

Print 1 to n using recursion

(Time Complexity is o(n))

```python
def printNumbers(i, n):
  # Base Case
  if i > n:
      return
  # Recursive Case
  print(i , end=" ")
  printNumbers(i+1, n) # recursion call


printNumbers(1,5)# Function call
```

#Factorial - A factorial is the product of an integer and all the positive whole numbers below it down to one. EX: 5! = 5 * 4 * 3 * 2 * 1

Using For loop

```python
n = int(input("Enter a number: "))

Factorial = 1

for i in range(1, n+1):
  Factorial *= i

print("Factorinal:", Factorial)
```

Using a function

```python
def factorial(n):
  fac = 1

  for i in range(1, n+1):
    fac *= i

  return fac

n = int(input("Enter a number: "))
print("Factorial:", factorial(n))
```

Using a recursion (Time Complexity is o(n))

```python
def factorial(n):

# base case
  if n == 0:
    return 1

# recursive case
  return n * factorial(n-1)

n = int(input("Enter the number:"))
print("Factorial:", factorial(n))
```

# 509 - Fibonacchi Number:
each no. is the sum of 2 preceding one

##Time Coomplexity is o(n²)

```python
class Solution:
  def fib(self, n: int) -> int:

    # base case
    if n == 0 or n == 1:
      return 1

    # recursive case
    return self.fib(n-1) + self.fib(n-2)

#create object:
s = Solution()
n = 5
print(s.fib(n))
```

#1137 Tribonacchi Series

Time Complexity o(n^3)

```python
class Solution:
    def tribonacci(self, n: int) -> int:

        if n == 0:
            return 0

        if n == 1 or n == 2:
            return 1

        return (
            self.tribonacci(n - 1)
            + self.tribonacci(n - 2)
            + self.tribonacci(n - 3)
        )

s = Solution()
print(s.tribonacci(5))
```

# 231 Power of 2 return true else false

using loop

```python
class Solution:
  def isPowerOftwo(self, n: int) ->bool:

    if n<=0:
      return False

    while n % 2 == 0:
      n //= 2

    return n == 1
s = Solution()
number = int(input("Enter the number: "))
print(s.isPowerOftwo(number))
```

using recursion

```python
class Solution:
  def isPowerOftwo(self, n: int) -> bool:

    #base case
    if n <0:
      return False
    if n == 1:
      return True
    if n%2!=0:
      return False

  #recursive case
    return self.isPowerOftwo(n//2)

s = Solution()
number = int(input("Enter the number: "))
print(s.isPowerOftwo(number))
```

Break both numbers down into their prime factors:

• 15 = $3 \times \mathbf{5}$

The only prime factor they share in common is 5. [4]

The least common multiple (**LCM**) of 15 and 50 is 150. [1]  

Here are **three different ways** to calculate this result:

**Method 1: Prime Factorization **

Find the prime factors of each number, then multiply the highest power of all prime numbers involved.

• Prime factors of 15: $3 \times 5 = 3^1 \times 5^1$

• Prime factors of 50: $2 \times 5 \times 5 = 2^1 \times 5^2$

• LCM calculation: Take the highest power of each prime factor ($2^1$, $3^1$, and $5^2$): $\text{LCM} = 2^1 \times 3^1 \times 5^2 = 2 \times 3 \times 25 = \mathbf{150}$  [2, 3]  

**Method 2: Listing Multiples **

List the multiples of each number until you find the first (smallest) one they share in common.

• Multiples of 15: 15, 30, 45, 60, 75, 90, 105, 120, 135, 150, 165...

• Multiples of 50: 50, 100, 150, 200... [2]  

The smallest shared multiple is 150. [2]  

**Method 3: The Division Method**

Divide both numbers together using prime numbers: [2]  

| Prime Divisor | Numbers  |
| --- | --- |
| 2 | 15, 50  |
| 3 | 15, 25  |
| 5 | 5, 25  |
| 5 | 1, 5  |
| — | 1, 1  |

Multiply the prime numbers on the left: $\text{LCM} = 2 \times 3 \times 5 \times 5 = \mathbf{150}$  [4]

#Two no. is equal to multiplication of gcd and lcm of 2 numbers

##**a * b = GCD(a,b) * LCM(a,b)**

# LCM of 2 numbers

LCM(a,b) = a*b / GCD(a,b)


```python
def gcd(a,b):

  if b == 0:
    return a
  return gcd(b, a%b)

def lcm(a,b):
  return a*b // gcd(a,b)

print("GCD/HCF: ",gcd(15,50))
print("LCM: ",lcm(15,50))
```

# 50. POW(x,n)

```python
class Solution:
  def findPow(self, x, n):
    #base case
    if n == 0:
      return 1
    #recursive case
    a = self.findPow(x, n//2)
    if n % 2 == 0:
      return a * a
    else:
      return x * a * a

  def myPow(self, x: float, n: int) -> float:
    if n>=0:
      return self.findPow(x, n)
    else:
      return 1/self.findPow(x, n*(-1))

s = Solution()
print(s.myPow(2,10))
```
