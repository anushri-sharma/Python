# FIZZBUZZ

```python
class fizzBuzzSolution:
    def fizzBuzz(self, n :int) -> list[str]:
        """

        means:

        Define a method called fizzBuzz that belongs to a class, takes an integer called n,
         and returns a list of strings.

        Here, "self" refers to the current Solution object.

        "n: int" ------ This is a type hint.

        "-> list[str]" -------This is the return type hint.
        It means: The function is expected to return a list containing strings.
            EX:   ["1", "2", "Fizz", "4", "Buzz"]

        """

        result = []
        for i in range(1, n+1):
            if i%3==0 and i%5==0:
                result.append("FIZZBUZZ")
            elif i%3==0:
                result.append("FIZZ")
            elif i%5==0:
                result.append("BUZZ")
            else:
                result.append(str(i))
        return result

# To test the solution, you would create an instance and call the method:

solution = fizzBuzzSolution() # fizzBuzzSolution needs to be an object (instance) of the class.
print(solution.fizzBuzz(20))
```

# 1365 - how many numbers are smaller than current number


```python
class solution3:
    def smallerNumbersThanCurrent(self, nums: list[int]) -> list[int]:
        result = []
        for i in nums:
            count = 0
            for j in nums:
                if j<i:
                    count += 1
            result.append(count)
        return result

solution = solution3()
print(solution.smallerNumbersThanCurrent([8,1,2,2,3]))
```

# 1281 - Subtract the product and sum of digits of an integers


```python
class solution:
    def subtractProductandsum(self, n: int) -> int:
        temp = n
        sum_ = 0
        product = 1

        while temp>0:
            # extract a digit with % 10, process it, then remove it with // 10.
            digit = temp%10
            temp //= 10
            sum_ += digit
            product *= digit
        return product - sum_
Solution = solution()
print(Solution.subtractProductandsum(234))
```

# 1431 - Kids with the greatest number of candies


```python
class solution:
    def kidswithCandies(self, candies: list[int], extraCandies: int) -> list[bool]:
        max_candies = max(candies)
        ans = []
        for i in candies:
            if i+extraCandies >= max_candies:
                ans.append(True)
            else:
                ans.append(False)
        return ans
Solution = solution()
print(Solution.kidswithCandies([2,3,5,1,3], 3))
```

```python
# by using list comprehension
class solution:
    def kidswithCandies(self, candies: list[int], extraCandies: int) -> list[bool]:
        maxcandies = max(candies)
        return [i + extraCandies >= maxcandies for i in candies]
Solution = solution()
print(Solution.kidswithCandies([2,3,5,1,3], 3))
```
